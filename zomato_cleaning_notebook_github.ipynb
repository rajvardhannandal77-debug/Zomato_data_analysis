# Zomato Data Cleaning

This notebook performs data cleaning and preprocessing on the Zomato restaurant dataset.

## Cleaning Steps
- Removed unnecessary columns
- Renamed columns for clarity
- Handled missing values
- Converted restaurant cost into numeric format
- Calculated cost per person
- Cleaned restaurant ratings
- Filled missing ratings using the mean rating
- Exported the cleaned dataset

> **Input:** `zomato.csv` (raw Zomato dataset)  
> **Output:** `zomato_clean_data.csv` (cleaned dataset)


# Import pandas library for data manipulation


```python
import pandas as pd
```

# Import numpy library for numerical operations


```python
import numpy as np
```

# Define a function 'handlerate' to clean the rating values


```python
def handlerate(value):
    # If value is 'NEW' or '-', return NaN
    if (value == 'NEW' or value == '-'):
        return np.nan
    else:
        # Otherwise, split by '/' and take the first part as float
        value = str(value).split('/')
        value = value[0]
        return float(value)
```

```python
# Read the raw Zomato dataset from a CSV file
df = pd.read_csv('zomato.csv')
```

```python
# Drop unnecessary columns from the dataframe
df3 = df.drop(columns=['url', 'phone', 'rest_type', 'dish_liked', 'reviews_list', 'menu_item', 'listed_in(city)'], axis=1)
```

```python
# Rename columns for better readability: cost and type columns
df4 = df3.rename(columns={'approx_cost(for two people)': 'two_people_cost', 'listed_in(type)': 'type_of_restaurant'})
```

```python
# Rename the 'rate' column to 'rating'
df5 = df4.rename(columns={'rate': 'rating'})
```

```python
# Drop rows with missing 'location' values
df5 = df5.dropna(subset=['location'])
```

```python
# Drop rows with missing 'cuisines' or 'two_people_cost' values
df5 = df5.dropna(subset=['cuisines', 'two_people_cost'])
```

```python
# Remove commas from 'two_people_cost' column and convert to integer
df5['two_people_cost'] = df5['two_people_cost'].str.replace(',', '').astype(int)
```

```python
# Calculate cost per person by dividing two_people_cost by 2
df5['two_people_cost'] = df5['two_people_cost'] / 2
```

```python
# Rename 'two_people_cost' column to 'cost_per_person'
df5 = df5.rename(columns={'two_people_cost': 'cost_per_person'})
```

```python
# Apply 'handlerate' function to clean the 'rating' column
df5['rating'] = df5['rating'].apply(handlerate)
```

```python
# Fill missing ratings with the mean rating
df5['rating'] = df5['rating'].fillna(df5['rating'].mean())
```

```python
# Save the cleaned dataframe to a new CSV file without the pandas index
df5.to_csv("zomato_clean_data.csv", index=False)
```
